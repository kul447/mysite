### Testing the Trained TFLite Model

Let's load the generated `currency_model_int8.tflite` model and perform a test inference on a sample image from the dataset. We'll pick a sample image and preprocess it to match the model's input requirements.

```python
import tensorflow as tf
import numpy as np
import os
import matplotlib.pyplot as plt

# Define the classes as used in training (7 classes)
CLASSES = ['10', '20', '50', '100', '200', '500', '2000']
IMG_SIZE = 128 # Image size used for training
DATA_DIR_CLEAN = '/content/data_clean'
MODEL_PATH = '/content/currency_model_int8.tflite'

# Load the TFLite model and allocate tensors.
interpreter = tf.lite.Interpreter(model_path=MODEL_PATH)
interpreter.allocate_tensors()

# Get input and output details
input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

print("Input details:", input_details)
print("Output details:", output_details)

def preprocess_image_for_inference(image_path, target_size, input_details):
    img = tf.keras.utils.load_img(image_path, target_size=target_size)
    img_array = tf.keras.utils.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) # Create a batch

    input_scale, input_zero_point = input_details[0]['quantization']

    if input_scale == 0.0 and input_zero_point == 0:
        input_data = (img_array / 127.5) - 1
        input_data = input_data.astype(np.float32)
    else:
        normalized_img_array = (img_array / 127.5) - 1
        input_data = (normalized_img_array / input_scale) + input_zero_point
        input_data = np.round(input_data).astype(input_details[0]['dtype'])
    return img, input_data

def run_inference(interpreter, input_data, input_details, output_details):
    interpreter.set_tensor(input_details[0]['index'], input_data)
    interpreter.invoke()
    output_data = interpreter.get_tensor(output_details[0]['index'])

    output_scale, output_zero_point = output_details[0]['quantization']
    if output_scale != 0.0 or output_zero_point != 0:
        output_data = output_scale * (output_data - output_zero_point)
    return output_data

# --- Test on multiple sample images ---
print("\n--- Testing Model with Sample Images ---")

num_samples_per_class = 2 # How many images to test per class

for class_label in CLASSES:
    sample_image_dir = os.path.join(DATA_DIR_CLEAN, class_label)
    if os.path.exists(sample_image_dir) and os.listdir(sample_image_dir):
        print(f"\nTesting class: {class_label}")
        images_in_class = os.listdir(sample_image_dir)
        np.random.shuffle(images_in_class) # Shuffle to get random samples

        for i, sample_image_name in enumerate(images_in_class):
            if i >= num_samples_per_class: break

            image_path = os.path.join(sample_image_dir, sample_image_name)
            img, input_data = preprocess_image_for_inference(image_path, (IMG_SIZE, IMG_SIZE), input_details)
            output_data = run_inference(interpreter, input_data, input_details, output_details)

            predicted_class_index = np.argmax(output_data)
            predicted_class_label = CLASSES[predicted_class_index]

            print(f"  Sample: {sample_image_name}")
            print(f"  Predicted: {predicted_class_label} (Actual: {class_label})")

            # Display the image with prediction
            plt.figure()
            plt.imshow(img)
            plt.title(f"Predicted: {predicted_class_label} (Actual: {class_label})")
            plt.axis('off')
            plt.show()

    else:
        print(f"Could not find sample images in {sample_image_dir} for testing.")
```
